```python
from pyspark.sql import SparkSession,DataFrame

spark = SparkSession.builder.appName("Read valid file").master("local").getOrCreate()
```

```python
df = spark.read.options(header = True,inferSchema=True,delimiter=",").csv("raw_files/orders_placed_20260925_231930.txt")
df.printSchema() 
```

```python
def read_valid_file(file_name:str,delimiter:str) -> DataFrame:
    """Read Valid File and Prepare DF to Dump in MySQL"""
    return  spark.read.options(header = True,inferSchema=True,delimiter=delimiter).csv(file_name)

        
read_valid_file("raw_files/orders_placed_20260925_231930.txt",delimiter=",")
    
```
