```python
import os

# create function to fetch all files in a targeted given dir
def fetch_target_files(base_dir:str, subfolder_name:str) -> list[str]:
    """Find and log all files in the specified incoming directory."""
    
    # create target directory path by using base_directory and subfolder name
    target_dir_path = os.path.join(base_dir,subfolder_name)

    # call out each files from the target dir path 
    try:
        return os.listdir(target_dir_path)
    except FileNotFoundError :
         print(f"Directory not found:{target_dir_path}")
         return []
        
# base directory for the incoming data pipeline
base_directory = 'C:\\Users\\v\\Multi Layer File Ingestion'

                
# run the function
all_files = fetch_target_files(base_directory,'raw_files')
```
