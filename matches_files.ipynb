```python
import re
```

```python
pattern = re.compile(r"^[a-zA-Z]+_[a-zA-Z]+_\d{8}_\d{6}.txt")

def pattern_matches_files(filename:str) -> bool:  
    """validate whether a filename matches <name>_<address>_<date>_<time>.txt format"""

    if not pattern.match(filename): 
        return False
    return True

pattern_matches_files("order_confirmed_12389768_894567.txt")    
```
