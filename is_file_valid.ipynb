```python
import csv
```

```python
def is_file_valid(file_name:str , column_length:int ,column_names:tuple):
    """validate whether a file has 4 columns and column names in ('order_id','customer_id','product_name','total_amount')  """
    with open(file_name,"r")as file:
        # first row take as column names
        header = csv.reader(file)
        column_name = next(header)
        
        if not len(column_name) == column_length and column_name not  in column_names:
            return True
        return False
            #print("File is valid and ready to dump into Mysql") 
```

```python

# run the function
is_file_valid('raw_files/customer_feedback_20260925_141522.txt',4,('order_id','customer_id','product_name','total_amount')) 
```
