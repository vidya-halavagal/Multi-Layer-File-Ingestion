# file_scanner: look at files in a folder 

## Import modules

```python
import os
```

## fetch_target_files
Return the list of files in the target directory.

```python
def fetch_target_files(target_dir: str) -> list[str]:
    """Fetch all files from the specified target directory.

    Args:
        target_dir: Path of the directory to scan.

    Returns:
        List of file names (folders are skipped). Empty list if the
        directory does not exist.
    """
    all_files = []

    # Check each item in the target directory
    try:
        for file_name in os.listdir(target_dir):
            full_file_path = os.path.join(target_dir, file_name)

            # Add only files, not folders
            if os.path.isfile(full_file_path):
                all_files.append(file_name)

        return all_files

    # Handle the case when the directory does not exist
    except FileNotFoundError:
        print(f"Directory not found: {target_dir}")
        return []
```

## is_unprocessed 
Return True if the file name is not already in the processed folder.

```python
def is_unprocessed(file_name: str, processed_folder: str) -> bool:
    """Check whether a file has not already been processed.

    Args:
        file_name: Name of the file to check.
        processed_folder: Path of the processed folder.

    Returns:
        True if the file name is not in the processed folder,
        otherwise False.
    """
    return file_name not in os.listdir(processed_folder)
```

## read_header_line 
Open a file and return its first line (the header).

```python
def read_header_line(file_path: str) -> str:
    """Read the first line (header) of a file.

    Args:
        file_path: Full path of the file to open.

    Returns:
        The first line of the file as a string.
    """
    with open(file_path, "r", newline="", encoding="utf-8") as file:
        header_line = file.readline().strip()
        return header_line
```
