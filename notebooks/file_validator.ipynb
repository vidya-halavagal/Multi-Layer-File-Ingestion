# file_validator: check file names and headers

## Import modules

```python
import re
```

## is_valid_file_name
Check a file name against the name pattern (name_address_date.txt). The pattern is passed in as a parameter.

```python
def is_valid_file_name(file_name: str, pattern: str) -> bool:
    """Check whether a file name matches <name>_<address>_<date>.txt format.

    Args:
        file_name: Name of the file to check.
        pattern: Regex pattern a valid file name must match.

    Returns:
        True if the whole file name matches the pattern, otherwise False.
    """
    file_pattern = re.compile(pattern)
    return file_pattern.fullmatch(file_name) is not None
```

```python
# test run
file_name_pattern = r"[a-zA-Z]+_[a-zA-Z]+_\d{8}\.txt"

print(is_valid_file_name("mary_paris_20261011.txt", file_name_pattern))   
print(is_valid_file_name("bad_name.txt", file_name_pattern))              
```

## split_files
Split a list of file names into valid and invalid lists.

```python
def split_files(file_names: list[str], pattern: str) -> tuple[list[str], list[str]]:
    """Split file names into valid and invalid lists based on a regex pattern.

    Args:
        file_names: List of file names to check.
        pattern: Regex pattern a valid file name must match.

    Returns:
        A tuple (valid_files, invalid_files) of two lists of file names.
    """
    valid_files = []
    invalid_files = []

    # sort file name into valid and invalid files 
    for file_name in file_names:
        if is_valid_file_name(file_name, pattern):
            # collect matched pattern file names into valid files list
            valid_files.append(file_name)
        else:
            # collect unmatched pattern file names into invalid files list
            invalid_files.append(file_name)
    
    return valid_files, invalid_files
```

```python
# test run

incoming_files = [
    "john_london_20261012.txt",
    "badname_file.txt",
    "mary_paris_20261011.txt",
    "notes.docx",
]

file_name_pattern = r"[a-zA-Z]+_[a-zA-Z]+_\d{8}\.txt"

valid_files, invalid_files = split_files(incoming_files, file_name_pattern)
print(valid_files)
print(invalid_files)
```

## has_valid_header
Check that the header line matches the expected columns: id, name, address, amount.

```python
def has_valid_header(header_line: str, expected_columns: list[str]) -> bool:
    """Check whether a header line has the expected columns in the expected order.

    Args:
        header_line: First line of a file, e.g. "id,name,address,amount".
        expected_columns: Column names the header must match, in order.

    Returns:
        True if the header matches, otherwise False.
    """
    columns = header_line.strip().split(",")
    return columns == expected_columns
```

```python
# test run
expected_columns = ["id", "name", "address", "amount"]

header_lines = [
    "id,name,address,amount",
    "id,fullname,city,price",
    "id,name,address",
]

for header_line in header_lines:
    is_valid = has_valid_header(header_line, expected_columns)
    print("Valid:" if is_valid else "Invalid:", header_line)
```
