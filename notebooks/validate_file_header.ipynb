### Validate File Header
read headers before reading the file

```python
from src.file_scanner import read_header_line
from src.file_validator import has_valid_header
```

```python
def validate_file_header(file_path: str, expected_cols: list[str] = None) -> bool:
    """Read the first line of the file and verify columns match expected list.

    Args:
        file_path: Full path to the target file.
        expected_cols: Expected column names. Defaults to ["id", "name", "address", "amount"].

    Returns:
        True if header matches expected columns, False otherwise.
    """
    file_header = read_header_line(file_path)
    
    if expected_cols is None:
        return has_valid_header(file_header)
    return has_valid_header(file_header,expected_cols)
```

```python
# test run
expected_columns = ["id", "name", "address", "amount"]

validate_file_header("../raw_files/john_london_20261012.txt",expected_cols=expected_columns)
```
