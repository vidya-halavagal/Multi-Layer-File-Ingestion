```python
import csv

def is_valid_file(filename: str, required_columns: tuple) -> bool:
    """Check whether a file contains the required columns in the expected order."""

    # Open the file and read its header
    with open(filename, "r") as file:
        reader = csv.reader(file)
        column_names = next(reader)

        # Compare the file columns with the required schema
        return (
            len(column_names) == len(required_columns)
            and tuple(column_names) == required_columns
        )
```

```python
#Testing with sample example
columns =('feedback_id', 'customer_id', 'rating', 'comments')

is_valid_file("raw_files/customer_feedback_20260925_141522.txt", columns)        
```
