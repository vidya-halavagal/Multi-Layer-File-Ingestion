# Validate file header 🧾
Check that the first line of a file has exactly the expected columns.

```python
def has_valid_header(header_line: str, expected_columns: list[str]) -> bool:
    """Check whether a header line has the expected columns in the expected order.

    Args:
        header_line: First line of a file, e.g. "id,name,address,amount".
        expected_columns: Column names the header must match, in order.

    Returns:
        True if the header matches, otherwise False.
    """
    columns = header_line.strip().split(",")
    return columns == expected_columns
```

```python
expected_columns = ["id", "name", "address", "amount"]

header_lines = [
    "id,name,address,amount",
    "id,fullname,city,price",
    "id,name,address",
]

for header_line in header_lines:
    is_valid = has_valid_header(header_line, expected_columns)
    print("Valid:" if is_valid else "Invalid:", header_line)
```
