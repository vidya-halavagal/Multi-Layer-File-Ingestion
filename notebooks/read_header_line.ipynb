# Read header line 
Open a file and return its first line (the header) 

```python
def read_header_line(file_path: str) -> str:
    """Read the first line (header) of a file.

    Args:
        file_path: Full path of the file to open.

    Returns:
        The first line of the file as a string.
    """
    with open(file_path, "r", newline="", encoding="utf-8") as file:
        header_line = file.readline().strip()
        return header_line
# test run 

read_header_line("C:/Users/v/Multi Layer File Ingestion/raw_files/john_london_20261012_093000.txt")
```
