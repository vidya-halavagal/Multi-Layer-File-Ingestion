### Your file names look like name_city_YYYYMMDD_HHMMSS.txt 

returns True only for names matching that pattern.

```python
def is_valid_file_name(file_name: str, pattern: str) -> bool:
    """Check whether a file name matches <name>_<address>_<date>_<time>.txt format.

    Args:
        file_name: Name of the file to check.
        pattern: Regex pattern a valid file name must match.

    Returns:
        True if the whole file name matches the pattern, otherwise False.
    """
    file_pattern = re.compile(pattern)
    return file_pattern.fullmatch(file_name) is not None
```

```python
#Testing with an example
my_pattern = r"[a-zA-Z]+_[a-zA-Z]+_\d{8}_\d{6}\.txt"
is_valid_file_name("mary_paris_20261011_141500.txt",my_pattern)    
```

### Split valid and invalid files 
Sort file names into valid 

and invalid  lists using the pattern

```python
def split_files(file_names: list[str], pattern: str) -> tuple[list[str], list[str]]:
    """Split file names into valid and invalid lists based on a regex pattern.

    Args:
        file_names: List of file names to check.
        pattern: Regex pattern a valid file name must match.

    Returns:
        A tuple (valid_files, invalid_files) of two lists of file names.
    """
    valid_files = []
    invalid_files = []

    # sort file name into valid and invalid files 
    for file_name in file_names:
        if is_valid_file_name(file_name, pattern):
            # collect matched pattern file names into valid files list
            valid_files.append(file_name)
        else:
            # collect unmatched pattern file names into invalid files list
            invalid_files.append(file_name)
    
    return valid_files, invalid_files
```

```python
#test run 
incoming_files = [
    "john_london_20261012_093000.txt",
    "badname_file.txt",
    "mary_paris_20261011_141500.txt",
    "notes.docx",
]

file_name_pattern = r"[a-zA-Z]+_[a-zA-Z]+_\d{8}_\d{6}\.txt"

valid_files, invalid_files = split_files(incoming_files,file_name_pattern) 
```
